```python
print("--- UNIVERSITY ADMISSION ELIGIBILITY ---")

course = input("Enter course (CS / Engineering / Medical): ")
marks = int(input("Enter your entry test marks: "))

if course == "CS":
    if marks >= 70:
        print("Congratulations! You are eligible for Computer Science.")
    if marks < 70:
        print("Under score! You need at least 70 marks for CS.")

elif course == "Engineering":
    if marks >= 80:
        print("Congratulations! You are eligible for Engineering.")
    if marks < 80:
        print("Under score! You need at least 80 marks for Engineering.")

elif course == "Medical":
    if marks >= 90:
        print("Congratulations! You are eligible for Medical.")
    if marks < 90:
        print("Under score! You need at least 90 marks for Medical.")

else:
    print("Invalid option. Please choose CS, Engineering, or Medical.")
```
